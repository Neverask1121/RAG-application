```python
import pandas as pd
```

```python
data = [100, 200, 300]
series = pd.Series(data, index=["a", "b", "c"])
series.loc["a"] = 200
print(series[series >= 200])
```

```python
calories = {
  "Day 1:":1750,
  "Day 2:":2100,
  "Day 3:":2000
}
series = pd.Series(calories)
series.loc["Day 3:"]+=500
print(series.loc["Day 3:"])
print(series[series > 1900])
```

```python
data = {
  "Name" :["Aditya", "Adrish", "Deepansh", "Sarthak"],
  "Age" :[18, 20, 18, 20],
}

df = pd.DataFrame(data, index=["Employee 1", "Employee 2", "Employee 3", "Employee 4"])
df["Job"] = ["Cook", "Serve", "Waiter", "Cleaner"]
new_row = pd.DataFrame([{"Name":"Shivang", "Age":22,  "Job":"Suck"}], index = ["Employee 5"])
df = pd.concat([df, new_row])
print(df)
print(df.iloc[2])
```
