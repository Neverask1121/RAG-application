```python
import os
from dotenv import load_dotenv
load_dotenv()
```

### We are going to use hugging face sentence transformers

```python
os.environ['HF_TOKEN'] = os.getenv('HF_TOKEN')
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings (model_name = "all-MiniLM-L6-v2") 
```

```python
embed_query = embeddings.embed_query("Hello world my name is Aditya")
embed_query
```

```python
len(embed_query)
```

```python
docs = embeddings.embed_documents(["I am about to blow", "Hello this is the first doc sentence"])
docs[0]
```
