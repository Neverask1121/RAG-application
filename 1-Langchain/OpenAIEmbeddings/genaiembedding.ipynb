```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings

# Correct: Use "text-embedding-004" (without "models/")
embeddings_1024 = GoogleGenerativeAIEmbeddings(model="text-embedding-004", dimensions=1024)

# Recommended: Update to the current embedding model
embeddings_1024 = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2-preview")
```

```python
text = "Hello my name is Aditya and please embed this"
result = embeddings_1024.embed_query(text)
result
```

```python
text = "Hello my name is Aditya and please embed this"
query_result = embeddings_1024.embed_query(text)
len(query_result)
```

```python
from langchain_community.document_loaders import TextLoader

loader=TextLoader('speech.txt')
docs=loader.load()
docs
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)
final_documents=text_splitter.split_documents(docs)
final_documents
```

```python
## Vector Embedding And Vector StoreDB
from langchain_community.vectorstores import Chroma

db=Chroma.from_documents(final_documents,embeddings_1024)
db

```

```python
### Retrieve the results from query vectorstore db
query="It will be all the easier for us to conduct ourselves as belligerents"
retrieved_results=db.similarity_search(query)
print(retrieved_results)
```
