```python
from langchain_text_splitters import HTMLHeaderTextSplitter

html_string = """
<!DOCTYPE html>
<html>
<body>
  <div>
    <h1>Fuck</h1>
    <p>Some intro text about why the fuck</p>
    <div>
      <h2>Bar main section</h2>
      <p>Hello there actually I was moaning</p>
      <h3>Bar section 1</h3>
      <p>Some text about the fucking of my wifu by me</p>
      <h3>Bar section 3</h3>
      <p>Some text about the satisfaction she got with me</p>
    </div>
    <div>
      <h2>Hello bro you good bro?</h2>
      <p>Why are you sharing sensitive information online bro wtf</p>
    </div>
    <br>
    <p>Here I enjoyed my night with her plus she wants more from me</p>
  </div>
</body>
</html>

"""

## html shit ends right here

headers_to_split_on = [
  ("h1", "Header 1"),
  ("h2", "Header 2"),
  ("h3", "Header 3")
]

html_splitter = HTMLHeaderTextSplitter(headers_to_split_on)
html_header_splits=html_splitter.split_text(html_string)
html_header_splits
```

```python
import requests
from langchain_text_splitters import HTMLHeaderTextSplitter

url = "https://plato.stanford.edu/"

response = requests.get(url)

html = response.text

headers_to_split_on = [
    ("h1", "Header 1"),
    ("h2", "Header 2"),
    ("h3", "Header 3"),
    ("h4", "Header 4"),
]

html_splitter = HTMLHeaderTextSplitter(headers_to_split_on)

docs = html_splitter.split_text(html)

docs
```
